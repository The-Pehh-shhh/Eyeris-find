# Eyeris-Find: Exploratory Data Analysis & Modeling

Developed a supervised machine learning classification engine to categorize Iris species based on sepal and petal morphometric dimensions. Handled data preprocessing, analyzed linear boundaries, and optimized predictive model accuracy across non-linearly separable feature classes to maximize model generalization.

## 1. Environment and Imports

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Project root
sys.path.insert(0, os.path.abspath('..'))

from src.data_loader import load_iris_data, audit_data, FEATURE_NAMES, TARGET_NAME, CLASS_LABELS
from src.predict import predict
```

## 2. Data Loading & Audit Findings
We load `bezdekIris.data`, which corrects two transcription errors present in UCI `iris.data`:
- Row 35: petal width corrected from 0.1 to 0.2
- Row 38: sepal width corrected from 3.1 to 3.6, petal length corrected from 1.5 to 1.4

```python
df = load_iris_data()
audit = audit_data(df)
print("Shape:", audit["shape"])
print("Missing values:", audit["total_missing"])
print("Duplicate rows:", audit["duplicate_rows"])
print("Class balance:", audit["class_balance"])
df.head()
```

## 3. Morphometric Distributions

```python
df.groupby(TARGET_NAME).describe().T
```

## 4. Linear Separability Analysis
Setosa is completely linearly separable from Versicolor and Virginica (100% linear accuracy).
Versicolor and Virginica overlap in feature space and require boundary optimization.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue=TARGET_NAME, style=TARGET_NAME, s=70)
plt.title('Petal Length vs Petal Width by Species', fontsize=13, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 5. Model Evaluation and Verification

```python
import json
metrics_path = os.path.join('..', 'models', 'metrics.json')
with open(metrics_path) as f:
    metrics = json.load(f)

print('Selected Best Model:', metrics['best_model'])
print(f"10-fold Repeated CV Accuracy: {metrics['cv_accuracy_mean']:.4f} +/- {metrics['cv_accuracy_std']:.4f}")
print(f"Held-out Test Accuracy: {metrics['test_accuracy']:.4f}")
print('Confusion Matrix:\n', np.array(metrics['confusion_matrix']))
```
